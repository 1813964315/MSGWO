```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator

plt.rcParams['font.family'] = 'Times New Roman'

# 假设有以下数据
x = [20, 30, 40, 50]  # x轴数据
# y = [1114.75, 1102.58, 1108.51, 1144.49]  # y轴数据，与x对应
# y = [481.59, 481.25, 487.24, 491.13]  # y轴数据，与x对应
# y = [924.83, 919.25, 922.93, 966.39]  # y轴数据，与x对应
y = [621.22, 595.02, 610.07, 622.74]  # y轴数据，与x对应
# 创建图表
plt.figure(figsize=(28, 16))

# 绘制折线图
plt.plot(x, y,color='black', linewidth=5)

# 设置x轴和y轴的刻度间隔
plt.gca().xaxis.set_major_locator(MultipleLocator(10))  # x轴刻度间隔为10

plt.tick_params(axis='x', labelsize=66)
plt.tick_params(axis='y', labelsize=66)
# 设置边框颜色和宽度
ax = plt.gca()  # 获取当前坐标轴
ax.spines['top'].set_color('black')  # 上边框颜色
ax.spines['right'].set_color('black')  # 右边框颜色
ax.spines['bottom'].set_color('black')  # 下边框颜色
ax.spines['left'].set_color('black')  # 左边框颜色
ax.spines['top'].set_linewidth(2)  # 上边框线宽
ax.spines['right'].set_linewidth(2)  # 右边框线宽
ax.spines['bottom'].set_linewidth(2)  # 下边框线宽
ax.spines['left'].set_linewidth(2)  # 左边框线宽
# 找到y数据的最小值及其索引
min_value = min(y)
min_index = y.index(min_value)

# 绘制并标注最小值的点，增加点的透明度
plt.scatter(x[min_index], min_value, color='red', s=800)
# 标签偏移量增加，垂直对齐方式始终在点上方
plt.text(x[min_index], min_value + 0.5, f'{min_value}', fontsize=55, color='black', ha='center', va='bottom', zorder=3)

# 绘制其余数据点，并在每个点旁边标注y值
for i, (xi, yi) in enumerate(zip(x, y)):
    if i != min_index:
        plt.scatter(xi, yi, color='black', s=800)  # 点的透明度
        # 根据点的y值动态调整偏移量和对齐方式
        text_offset = -3  # 增加偏移量
        plt.text(xi, yi + text_offset, f'{yi}', fontsize=55, color='black', ha='center', va='bottom', zorder=3)

# 添加图例
# plt.legend(fontsize=48)

# 添加标题和轴标签
plt.xlabel('Population Size', fontsize=66)
plt.ylabel('Total Distance (km)', fontsize=66)

# 保存图表
# plt.savefig('1.png', dpi=600)
# plt.savefig('2.png', dpi=300)
# plt.savefig('3.png', dpi=300)
plt.savefig('4.png', dpi=300)

# 显示图表
plt.show()
```
