```python
import matplotlib.pyplot as plt
import numpy as np

# Set font to Times New Roman
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams.update({'font.size': 14})

# Define the data directly as arrays
algorithms = ['PGWO', 'EGWO', 'MSGWO']
datasets = ['CPS01', 'CPS02', 'CPS03', 'CPS04']

# Total distances
total_distances = [
    [1188.858, 491.125, 949.922, 633.205],  # PGWO
    [1272.430, 567.034, 1026.279, 609.361], # EGWO
    [1121.079, 477.877, 930.483, 582.158]   # MSGWO
]

# Average distances
average_distances = [
    [113.063, 144.067, 132.986, 135.969],   # PGWO
    [121.011, 166.334, 143.676, 130.848],   # EGWO
    [106.617, 140.181, 130.265, 125.007]    # MSGWO
]

# Plotting the bar chart
fig, ax = plt.subplots(figsize=(14, 10))

# Define the positions for the groups of bars with additional spacing
bar_width = 0.1
spacing = 0.1
index = np.arange(len(datasets)) * (len(algorithms) * 2 * bar_width + spacing)

# Define colors for each algorithm
colors = ['#FDC897', '#ECA8A9', '#D3E2B7']

# Plot bars for each dataset
for j, dataset in enumerate(datasets):
    for i, algorithm in enumerate(algorithms):
        # Plot total distances
        plt.bar(index[j] + i * bar_width, 
                total_distances[i][j], 
                bar_width, 
                color=colors[i],
                label=f'{algorithm} Total' if j == 0 else "")  # Add label only once for the legend
        # Plot average distances
        plt.bar(index[j] + (i + len(algorithms)) * bar_width, 
                average_distances[i][j], 
                bar_width, 
                color=colors[i],
                hatch='//',
                label=f'{algorithm} Average' if j == 0 else "")  # Add label only once for the legend

# Set the x-axis labels and ticks
ax.set_xlabel('Datasets', fontsize=26)
ax.set_ylabel('Distance(km)', fontsize=26)
# ax.set_title('Comparison of Total and Average Distances for Different Algorithms across Datasets', fontsize=20)
ax.set_xticks(index + bar_width * 1.5)
ax.set_xticklabels(datasets, fontsize=26)
ax.tick_params(axis='y', labelsize=26)

# Add a legend with custom handles
from matplotlib.patches import Patch

legend_handles = [
    Patch(facecolor='#FDC897', edgecolor='#FDC897', label='PGWO Total'),
    Patch(facecolor='#FDC897', edgecolor='#FDC897',label='PGWO Average', hatch='//'),
    Patch(facecolor='#ECA8A9', edgecolor='#ECA8A9', label='EGWO Total'),
    Patch(facecolor='#ECA8A9',edgecolor='#ECA8A9', label='EGWO Average', hatch='//'),
    Patch(facecolor='#D3E2B7', edgecolor='#D3E2B7', label='MSGWO Total'),
    Patch(facecolor='#D3E2B7',edgecolor='#D3E2B7', label='MSGWO Average', hatch='//')
]

plt.legend(handles=legend_handles, fontsize=14)
plt.savefig('celue.png', dpi=96, bbox_inches='tight')
# Show the plot
plt.show()

```
