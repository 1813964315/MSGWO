```python
import numpy as np
import matplotlib.pyplot as plt

def plot_lines_with_larger_fonts_and_thicker_lines(curves, labels, colors, final_curve, final_label, final_color):
    plt.figure(figsize=(12, 8))

    for curve, label, color in zip(curves, labels, colors):
        plt.plot(curve, label=label, color=color, linewidth=3)  # 增加线条粗细

    # Plot the final curve with a different style
    plt.plot(final_curve, label=final_label, color=final_color, linewidth=3.5)  # 增加最终线条的粗细

    plt.xlabel('Iteration', fontsize=18)
    plt.ylabel('Objective Function Value', fontsize=18)
    plt.title('Convergence Curve', fontsize=20)
    plt.legend(fontsize=16)
    # plt.grid(True)
    plt.xticks(fontsize=16)
    plt.yticks(fontsize=16)
    plt.show()

# 使用示例数据运行函数
curves = [convergence_curve1, convergence_curve2, convergence_curve3, convergence_curve4, convergence_curve5, convergence_curve6, convergence_curve7]
labels = ['GWO', 'HO', 'APO', 'BSLO', 'IVY', 'RBMO', 'WOA']
colors = ['blue', 'green', 'orange', 'purple', 'brown', 'pink', 'gray']
final_curve = convergence_curve
final_label = 'MSGWO'
final_color = 'red'

plot_lines_with_larger_fonts_and_thicker_lines(curves, labels, colors, final_curve, final_label, final_color)
```
